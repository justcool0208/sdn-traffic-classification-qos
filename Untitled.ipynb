```python
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split
```

```python
datasets = {
    "minmax": df_minmax,
    "robust": df_robust,
    "std": df_std,
}

# Requested train-test split settings (test size)
splits = {
    "70-30": 0.30,
    "60-40": 0.40,
    "50-50": 0.50,
    "30-70": 0.70,
}

split_data = {}
```

```python
for dataset_name, frame in datasets.items():
    X = frame.drop(["Category", "binary_category"], axis=1).copy()
    y = frame["binary_category"]

    non_numeric_cols = X.select_dtypes(include=["object", "string", "category"]).columns
    for col in non_numeric_cols:
        X[col], _ = pd.factorize(X[col])

    split_data[dataset_name] = {}
    for split_name, test_size in splits.items():
        X_train, X_test, y_train, y_test = train_test_split(
            X,
            y,
            test_size=test_size,
            random_state=42,
            stratify=y,
        )

        split_data[dataset_name][split_name] = {
            "X_train": X_train,
            "X_test": X_test,
            "y_train": y_train,
            "y_test": y_test,
        }
```

```python
smote_outputs = {}

for dataset_name, split_map in split_data.items():
    smote_outputs[dataset_name] = {}

    for split_name, split_vals in split_map.items():
        y_train = split_vals["y_train"]
        class_counts = y_train.value_counts()

        if class_counts.shape[0] < 2:
            smote_outputs[dataset_name][split_name] = {
                "X_train_res": split_vals["X_train"],
                "y_train_res": y_train,
                "smote_applied": False,
                "reason": "only one class in y_train",
            }
            continue

        minority_count = class_counts.min()
        if minority_count < 2:
            smote_outputs[dataset_name][split_name] = {
                "X_train_res": split_vals["X_train"],
                "y_train_res": y_train,
                "smote_applied": False,
                "reason": "minority class has fewer than 2 samples",
            }
            continue

        k_neighbors = min(5, minority_count - 1)
        smote = SMOTE(random_state=42, k_neighbors=k_neighbors)
        X_train_res, y_train_res = smote.fit_resample(split_vals["X_train"], y_train)

        smote_outputs[dataset_name][split_name] = {
            "X_train_res": X_train_res,
            "y_train_res": y_train_res,
            "smote_applied": True,
            "reason": "ok",
        }
```

```python
for dataset_name in datasets:
    print(f"\n{'='*60}")
    print(f"Dataset: {dataset_name.upper()}")
    print(f"{'='*60}")

    for split_name in splits:
        print(f"\nSplit: {split_name} (test={int(splits[split_name] * 100)}%)")
        print("Original train class counts:")
        orig_counts = split_data[dataset_name][split_name]["y_train"].value_counts().sort_index()
        print(f"  STREAMING (0): {orig_counts.get(0, 0)}")
        print(f"  Others (1):    {orig_counts.get(1, 0)}")

        print("After SMOTE balancing:")
        smote_counts = smote_outputs[dataset_name][split_name]["y_train_res"].value_counts().sort_index()
        print(f"  STREAMING (0): {smote_counts.get(0, 0)}")
        print(f"  Others (1):    {smote_counts.get(1, 0)}")

        print(f"SMOTE applied: {smote_outputs[dataset_name][split_name]['smote_applied']}")
        if not smote_outputs[dataset_name][split_name]["smote_applied"]:
            print(f"Reason: {smote_outputs[dataset_name][split_name]['reason']}")
```

```python
from sklearn.base import clone
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    cohen_kappa_score,
    confusion_matrix,
    matthews_corrcoef,
    )
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import VotingClassifier

xgb_available = True
try:
    from xgboost import XGBClassifier
except ImportError:
    xgb_available = False
    print("xgboost not installed. Running DT, KNN, SVM, and Hybrid models.")
```

```python
train_data = {}
for dataset_name in datasets:
    train_data[dataset_name] = {}
    for split_name in splits:
        train_data[dataset_name][split_name] = {
            "X_train": smote_outputs[dataset_name][split_name]["X_train_res"],
            "y_train": smote_outputs[dataset_name][split_name]["y_train_res"],
            "X_test": split_data[dataset_name][split_name]["X_test"],
            "y_test": split_data[dataset_name][split_name]["y_test"],
        }

# Fast-model settings
FAST_MODE = True
FAST_XGB_ESTIMATORS = 60

# Train on full training split by default.
# Set to an integer (e.g., 25000) only if you want optional SVM subsampling.
MAX_SVM_SAMPLES = None

# Reusable base estimators
base_dt = DecisionTreeClassifier(
    random_state=42,
    max_depth=8,
    min_samples_leaf=3,
    min_samples_split=10,
    class_weight="balanced",
)
base_knn = KNeighborsClassifier(
    n_neighbors=7,
    weights="distance",
    n_jobs=-1,
)
base_svm = LinearSVC(
    C=1.0,
    max_iter=2000,
    random_state=42,
    class_weight="balanced",
    dual="auto",
)

models = {
    "DT": base_dt,
    "KNN": base_knn,
    "SVM": base_svm,
    "Hybrid DT-SVM": VotingClassifier(
        estimators=[
            ("svm", clone(base_svm)),
            ("dt", clone(base_dt)),
        ],
        voting="hard",
        n_jobs=-1,
    ),
    "Hybrid KNN-SVM": VotingClassifier(
        estimators=[
            ("svm", clone(base_svm)),
            ("knn", clone(base_knn)),
        ],
        voting="hard",
        n_jobs=-1,
    ),
}

if xgb_available:
    models["XGBoost"] = XGBClassifier(
        n_estimators=FAST_XGB_ESTIMATORS if FAST_MODE else 120,
        max_depth=4,
        learning_rate=0.1,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1,
        tree_method="hist",
        verbosity=0,
    )
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display

results = []
trained_models = {}
confusion_matrices = {}

total_iterations = len(datasets) * len(splits) * len(models)
current_iteration = 0

print(f"Training {len(models)} models on {len(datasets)} datasets and {len(splits)} splits...")
print(f"Total: {total_iterations} model trainings\n")

def maybe_subsample_for_svm(X_train, y_train, model_name, max_samples=None):
    svm_like = {"SVM", "Hybrid DT-SVM", "Hybrid KNN-SVM"}
    if model_name not in svm_like:
        return X_train, y_train, False

    if max_samples is None or max_samples <= 0 or len(y_train) <= max_samples:
        return X_train, y_train, False

    rng = np.random.default_rng(42)
    indices = rng.choice(len(y_train), size=max_samples, replace=False)

    if hasattr(X_train, "iloc"):
        X_sub = X_train.iloc[indices]
    else:
        X_sub = X_train[indices]

    if hasattr(y_train, "iloc"):
        y_sub = y_train.iloc[indices]
    else:
        y_sub = y_train[indices]

    return X_sub, y_sub, True

for dataset_name, split_map in train_data.items():
    trained_models[dataset_name] = {}

    for split_name, data in split_map.items():
        X_train = data["X_train"]
        y_train = data["y_train"]
        X_test = data["X_test"]
        y_test = data["y_test"]

        trained_models[dataset_name][split_name] = {}

        for model_name, model in models.items():
            current_iteration += 1
            start_time = time.time()

            print(
                f"[{current_iteration}/{total_iterations}] Training {model_name} on {dataset_name} ({split_name})...",
                end=" ",
            )

            clf = clone(model)
            if hasattr(clf, "n_jobs") and model_name != "XGBoost":
                clf.n_jobs = -1

            X_fit, y_fit, subsampled = maybe_subsample_for_svm(
                X_train,
                y_train,
                model_name,
                max_samples=MAX_SVM_SAMPLES if "MAX_SVM_SAMPLES" in globals() else None,
            )

            clf.fit(X_fit, y_fit)
            y_pred = clf.predict(X_test)

            elapsed = time.time() - start_time
            note = " [subsampled]" if subsampled else ""
            print(f"OK ({elapsed:.2f}s){note}")

            cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
            confusion_matrices[(dataset_name, split_name, model_name)] = cm

            results.append({
                "dataset": dataset_name,
                "split": split_name,
                "model": model_name,
                "accuracy": accuracy_score(y_test, y_pred),
                "precision": precision_score(y_test, y_pred, zero_division=0),
                "recall": recall_score(y_test, y_pred, zero_division=0),
                "f1": f1_score(y_test, y_pred, zero_division=0),
                "kappa": cohen_kappa_score(y_test, y_pred),
                "mcc": matthews_corrcoef(y_test, y_pred),
                "training_time": elapsed,
                "subsampled": subsampled,
            })

            trained_models[dataset_name][split_name][model_name] = clf

print("\n" + "=" * 60)
print("Training complete!")
print("=" * 60 + "\n")

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(["dataset", "split", "f1"], ascending=[True, True, False]).reset_index(drop=True)

print("Model performance by dataset and split:")
display(results_df)

split_order = list(splits.keys())
model_order = [name for name in ["DT", "Hybrid DT-SVM", "Hybrid KNN-SVM", "KNN", "SVM", "XGBoost"] if name in models]

for dataset_name in datasets:
    print(f"\n{'=' * 60}")
    print(f"Summary table for dataset: {dataset_name.upper()}")
    print(f"{'=' * 60}")

    ds = results_df[results_df["dataset"] == dataset_name].copy()

    table_accuracy = ds.pivot_table(index="split", columns="model", values="accuracy")
    table_accuracy = table_accuracy.reindex(split_order)
    table_accuracy = table_accuracy.reindex(columns=model_order)
    table_accuracy_pct = (table_accuracy * 100).round(2)
    table_accuracy_pct.index = [f"test = {int(splits[s] * 100)}%" for s in table_accuracy_pct.index]

    print("\nAccuracy table (%)")
    display(table_accuracy_pct)

    print("\nConfusion matrix for each model and split")
    for split_name in split_order:
        print(f"\nDataset: {dataset_name.upper()} | Split: {split_name}")
        for model_name in model_order:
            key = (dataset_name, split_name, model_name)
            if key not in confusion_matrices:
                continue
            cm = confusion_matrices[key]
            print(f"\nModel: {model_name}")
            display(pd.DataFrame(cm, index=["Actual 0", "Actual 1"], columns=["Pred 0", "Pred 1"]))

    # Graphs similar to your example: Accuracy/F1/Kappa by model for each split
    fig, axes = plt.subplots(2, 2, figsize=(18, 11), constrained_layout=True)
    axes = axes.flatten()

    for idx, split_name in enumerate(split_order):
        ax = axes[idx]
        split_rows = ds[ds["split"] == split_name].copy()
        split_rows = split_rows.set_index("model").reindex(model_order).reset_index()

        x = np.arange(len(model_order))
        width = 0.25

        acc_vals = (split_rows["accuracy"].fillna(0).values * 100)
        f1_vals = (split_rows["f1"].fillna(0).values * 100)
        kappa_vals = (split_rows["kappa"].fillna(0).values * 100)

        ax.bar(x - width, acc_vals, width=width, label="Accuracy", color="blue")
        ax.bar(x, f1_vals, width=width, label="F1 Score", color="green")
        ax.bar(x + width, kappa_vals, width=width, label="Kappa Score", color="orange")

        ax.set_title(f"{dataset_name.upper()} | Split {split_name}")
        ax.set_xticks(x)
        ax.set_xticklabels(model_order, rotation=30, ha="right")
        ax.set_ylim(0, 105)
        ax.set_ylabel("Score (%)")
        if idx == 0:
            ax.legend()

    fig.suptitle(
        f"Performance comparison of ML models for binary classification ({dataset_name.upper()})",
        fontsize=14,
    )
    plt.show()

results_df
```
